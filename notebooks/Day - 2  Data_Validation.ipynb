```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

## Loading cleaned dataset from Day 1

```python
day1_path = "../data/cleaned_data/APL_Day1_Cleaned.csv"

df_day1 = pd.read_csv(
    day1_path,
    encoding="cp1252",
    low_memory=False
)

print("Day 1 cleaned dataset loaded successfully.")
print("Shape:", df_day1.shape)
```

## Creating Day 2 working copy 

```python
df_day2 = df_day1.copy()

print("Day 2 validation copy created.")
print("Shape:", df_day2.shape)
```

## Validation 1 - Dataset shape 

```python
expected_rows = 180519
expected_columns = 40

actual_rows, actual_columns = df_day2.shape

print("Expected rows:", expected_rows)
print("Actual rows:", actual_rows)

print("Expected columns:", expected_columns)
print("Actual columns:", actual_columns)

if (actual_rows == expected_rows) and (actual_columns == expected_columns):
    print("Shape validation: PASS")
else:
    print("Shape validation: CHECK REQUIRED")
```

## Validation 2 - Columns names

```python
expected_columns = [
    "Type",
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Benefit per order",
    "Sales per customer",
    "Delivery Status",
    "Late_delivery_risk",
    "Category Id",
    "Category Name",
    "Customer City",
    "Customer Country",
    "Customer Fname",
    "Customer Id",
    "Customer Lname",
    "Customer Segment",
    "Customer State",
    "Customer Street",
    "Customer Zipcode",
    "Department Id",
    "Department Name",
    "Latitude",
    "Longitude",
    "Market",
    "Order City",
    "Order Country",
    "Order Customer Id",
    "Order Item Discount",
    "Order Item Discount Rate",
    "Order Item Product Price",
    "Order Item Profit Ratio",
    "Order Item Quantity",
    "Sales",
    "Order Item Total",
    "Order Profit Per Order",
    "Order Region",
    "Order State",
    "Order Status",
    "Product Name",
    "Product Price",
    "Shipping Mode"
]

actual_columns = df_day2.columns.tolist()

missing_columns = set(expected_columns) - set(actual_columns)
unexpected_columns = set(actual_columns) - set(expected_columns)

print("Missing columns:", missing_columns)
print("Unexpected columns:", unexpected_columns)

if not missing_columns and not unexpected_columns:
    print("Column validation: PASS")
else:
    print("Column validation: CHECK REQUIRED")
```

## Validation 3 - Missing values

```python
missing_values = df_day2.isnull().sum()

missing_summary = (
    missing_values[missing_values > 0]
    .sort_values(ascending=False)
)

display(missing_summary)

if missing_summary.empty:
    print("Missing value validation: PASS")
else:
    print("Missing value validation: CHECK REQUIRED")
```

## Validation 4 - Duplicate values

```python
duplicate_count = df_day2.duplicated().sum()

print("Duplicate rows:", duplicate_count)

if duplicate_count == 0:
    print("Duplicate validation: PASS")
else:
    print("Duplicate validation: CHECK REQUIRED")
```

## Validation 5 - Data types

```python
dtype_summary = pd.DataFrame({
    "Column": df_day2.columns,
    "Data_Type": df_day2.dtypes.astype(str).values
})

display(dtype_summary)
```

## Validation 6 - Categorical vlaues

```python
categorical_columns = [
    "Type",
    "Delivery Status",
    "Customer Segment",
    "Market",
    "Order Status",
    "Shipping Mode"
]

for col in categorical_columns:
    print("\n" + "=" * 60)
    print(col)
    print("=" * 60)
    print(df_day2[col].value_counts())
```

## Validation 7 - Late delivery risk

```python
valid_risk_values = {0, 1}

actual_risk_values = set(
    df_day2["Late_delivery_risk"].dropna().unique()
)

print("Actual values:", actual_risk_values)

if actual_risk_values.issubset(valid_risk_values):
    print("Late_delivery_risk validation: PASS")
else:
    print("Late_delivery_risk validation: CHECK REQUIRED")
```

## Validation 8 - Shipping days

```python
shipping_columns = [
    "Days for shipping (real)",
    "Days for shipment (scheduled)"
]

for col in shipping_columns:
    print("\n", col)
    print(sorted(df_day2[col].dropna().unique()))
```

```python
valid_real_days = set(range(0, 7))
valid_scheduled_days = {0, 1, 2, 4}

actual_real_days = set(
    df_day2["Days for shipping (real)"].dropna().unique()
)

actual_scheduled_days = set(
    df_day2["Days for shipment (scheduled)"].dropna().unique()
)

print(
    "Real shipping days:",
    actual_real_days.issubset(valid_real_days)
)

print(
    "Scheduled shipping days:",
    actual_scheduled_days.issubset(valid_scheduled_days)
)
```

## Validation 9 - Delivery logic 

```python
expected_late = (
    (df_day2["Days for shipping (real)"] >
     df_day2["Days for shipment (scheduled)"])
    & (df_day2["Delivery Status"] != "Shipping canceled")
)

actual_late = df_day2["Delivery Status"] == "Late delivery"

logic_mismatch = df_day2[expected_late != actual_late]

print("Expected late deliveries:", expected_late.sum())
print("Actual late deliveries:", actual_late.sum())
print("Delivery logic mismatches:", len(logic_mismatch))
```

```python
display(
    logic_mismatch[
        [
            "Days for shipping (real)",
            "Days for shipment (scheduled)",
            "Delivery Status",
            "Late_delivery_risk"
        ]
    ].head(20)
)
```

## Validation 10 - Late risk vs delivery status 

```python
risk_mismatch = df_day2[
    (
        (df_day2["Delivery Status"] == "Late delivery") &
        (df_day2["Late_delivery_risk"] != 1)
    )
    |
    (
        (df_day2["Delivery Status"] != "Late delivery") &
        (df_day2["Late_delivery_risk"] != 0)
    )
]

print("Risk/status mismatches:", len(risk_mismatch))
```

## Validation 11 - Numeric range check 

```python
numeric_columns = df_day2.select_dtypes(
    include=["int64", "float64"]
).columns

numeric_summary = df_day2[numeric_columns].describe().T

display(numeric_summary)
```

### Quantity

```python
print(
    "Invalid quantities:",
    (df_day2["Order Item Quantity"] <= 0).sum()
)
```

### Discount rate

```python
print(
    "Invalid discount rates:",
    (
        (df_day2["Order Item Discount Rate"] < 0) |
        (df_day2["Order Item Discount Rate"] > 1)
    ).sum()
)
```

## Latitude 

```python
print(
    "Invalid latitude:",
    (
        (df_day2["Latitude"] < -90) |
        (df_day2["Latitude"] > 90)
    ).sum()
)
```

### Longitude 

```python
print(
    "Invalid longitude:",
    (
        (df_day2["Longitude"] < -180) |
        (df_day2["Longitude"] > 180)
    ).sum()
)
```

### Validation 12 - Price consistency

```python
print(
    "Invalid Product Price:",
    (df_day2["Product Price"] <= 0).sum()
)

print(
    "Invalid Order Item Product Price:",
    (df_day2["Order Item Product Price"] <= 0).sum()
)
```

## Validation 13 - Financial consistency

```python
financial_columns = [
    "Benefit per order",
    "Sales per customer",
    "Order Item Discount",
    "Order Item Discount Rate",
    "Order Item Product Price",
    "Order Item Profit Ratio",
    "Sales",
    "Order Item Total",
    "Order Profit Per Order",
    "Product Price"
]

display(df_day2[financial_columns].describe().T)
```

## Validation 14 - ID consistency

```python
id_columns = [
    "Category Id",
    "Customer Id",
    "Department Id",
    "Order Customer Id"
]

for col in id_columns:
    print(
        col,
        "unique values:",
        df_day2[col].nunique(),
        "| missing:",
        df_day2[col].isna().sum()
    )
```

```python
print(
    "Customer Id < 0:",
    (df_day2["Customer Id"] < 0).sum()
)

print(
    "Category Id <= 0:",
    (df_day2["Category Id"] <= 0).sum()
)

print(
    "Department Id <= 0:",
    (df_day2["Department Id"] <= 0).sum()
)
```

## Validation 15 - Customer ID relationship 

```python
customer_id_mismatch = (
    df_day2["Customer Id"] !=
    df_day2["Order Customer Id"]
).sum()

print(
    "Customer ID mismatches:",
    customer_id_mismatch
)
```

## Validation 16 - Text quality

```python
text_columns = df_day2.select_dtypes(
    include=["object", "string"]
).columns

whitespace_issues = {}

for col in text_columns:
    count = (
        df_day2[col].astype("string")
        .str.contains(r"^\s|\s$", regex=True, na=False)
        .sum()
    )

    if count > 0:
        whitespace_issues[col] = count

print("Whitespace issues:")
print(whitespace_issues)
```

## Creating a Day 2 validation summary

```python
validation_results = {

    "Shape": df_day2.shape == (180519, 40),

    "Missing Values": df_day2.isnull().sum().sum() == 0,

    "Duplicate Rows": df_day2.duplicated().sum() == 0,

    "Required Columns": set(expected_columns) == set(df_day2.columns),

    "Late Delivery Risk": set(
        df_day2["Late_delivery_risk"].dropna().unique()
    ).issubset({0, 1}),

    "Shipping Days": actual_real_days.issubset(valid_real_days),

    "Scheduled Days": actual_scheduled_days.issubset(valid_scheduled_days),

    "Delivery Logic": len(logic_mismatch) == 0,

    "Positive Quantity": (df_day2["Order Item Quantity"] > 0).all(),

    "Valid Latitude": (
        df_day2["Latitude"].between(-90, 90)
    ).all(),

    "Valid Longitude": (
        df_day2["Longitude"].between(-180, 180)
    ).all()
}

validation_report = pd.DataFrame(
    list(validation_results.items()),
    columns=["Validation", "Status"]
)

display(validation_report)
```

## Overall Day 2 result

```python
if validation_report["Status"].all():
    print("DAY 2 DATA VALIDATION: PASS")
else:
    print("DAY 2 DATA VALIDATION: CHECK REQUIRED")
```

## Saving Day 2 dataset

```python
day2_path = "../data/cleaned_data/APL_Day2_Validated.csv"

df_day2.to_csv(
    day2_path,
    index=False,
    encoding="cp1252"
)

print("Day 2 validated dataset saved successfully.")
print(day2_path)
```

```python
validation_report_path = "../data/cleaned_data/APL_Day2_Validation_Report.csv"

validation_report.to_csv(
    validation_report_path,
    index=False,
    encoding="utf-8"
)

print("Validation report saved successfully.")
print(validation_report_path)
```
