```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

## Load canonical dataset

```python
import pandas as pd

raw_path = "../data/raw_data/APL_Logistics.csv"

df = pd.read_csv(
    raw_path,
    encoding="cp1252"
)

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

## Inspect the dataset

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())
```

## Check the data types

```python
df.info()
```

```python
dtype_summary = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values
})

display(dtype_summary)
```

## Missing value analysis

```python
missing = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isnull().sum().values,
    "Missing_Percentage": (
        df.isnull().mean().values * 100
    )
})

missing = missing.sort_values(
    "Missing_Count",
    ascending=False
)

display(missing[missing["Missing_Count"] > 0])
```

## Duplicate check

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

## Checking unique values of important categorical fields

```python
categorical_columns = [
    "Type",
    "Delivery Status",
    "Customer Segment",
    "Market",
    "Order Status",
    "Shipping Mode"
]

for col in categorical_columns:
    print("\n" + "=" * 60)
    print(col)
    print("=" * 60)
    print(df[col].value_counts(dropna=False))
```

## Validate the shipping duration fields

```python
shipping_columns = [
    "Days for shipping (real)",
    "Days for shipment (scheduled)"
]

display(df[shipping_columns].describe())
```

```python
for col in shipping_columns:
    print("\n", col)
    print(sorted(df[col].dropna().unique()))
```

## Validate (Late delivery risk)

```python
print(df["Late_delivery_risk"].value_counts())

print("\nPercentage:")
print(
    df["Late_delivery_risk"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
valid_risk_values = {0, 1}

actual_risk_values = set(
    df["Late_delivery_risk"].dropna().unique()
)

print("Risk values:", actual_risk_values)

if actual_risk_values.issubset(valid_risk_values):
    print("Late_delivery_risk validation: PASS")
else:
    print("Late_delivery_risk validation: CHECK REQUIRED")
```

## Validate delivery status

```python
print(df["Delivery Status"].value_counts())
```

## Basic numeric validation

```python
numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

numeric_summary = df[numeric_columns].describe().T

display(numeric_summary)
```

```python
checks = {
    "Negative shipping days":
        (df["Days for shipping (real)"] < 0).sum(),

    "Negative scheduled days":
        (df["Days for shipment (scheduled)"] < 0).sum(),

    "Invalid late risk":
        (~df["Late_delivery_risk"].isin([0, 1])).sum(),

    "Negative quantity":
        (df["Order Item Quantity"] <= 0).sum(),

    "Negative discount":
        (df["Order Item Discount"] < 0).sum(),

    "Negative discount rate":
        (df["Order Item Discount Rate"] < 0).sum()
}

for check, count in checks.items():
    print(f"{check}: {count}")
```

## Handle missing values carefully

```python
df["Customer Lname"] = df["Customer Lname"].fillna("Unknown")
df["Customer Zipcode"] = df["Customer Zipcode"].fillna("Unknown")
```

## Standardize text fields

```python
# Strip leading/trailing whitespace from text columns
# Exclude Customer Zipcode because it contains numeric values

text_columns = df.select_dtypes(
    include=["object", "string"]
).columns

text_columns = text_columns.drop(
    "Customer Zipcode",
    errors="ignore"
)

for col in text_columns:
    df[col] = df[col].str.strip()
```

## Check the final Day 1 quality

```python
print("Final Shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

## Saving the cleaned dataset

```python
cleaned_path = "../data/cleaned_data/APL_Day1_Cleaned.csv"

df.to_csv(
    cleaned_path,
    index=False,
    encoding="cp1252"
)

print("Day 1 cleaned dataset saved successfully.")
print(cleaned_path)
```
